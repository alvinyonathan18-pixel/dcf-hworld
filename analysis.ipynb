```python
base_revenue = income_stmt.loc["Total Revenue"]["2025-12-31"]

scenarios = {
    "Bear": {"growth": 0.02, "fcf_margin": 0.26},
    "Base": {"growth": 0.06, "fcf_margin": 0.29},
    "Bull": {"growth": 0.10, "fcf_margin": 0.31},
}

years = [1, 2, 3, 4, 5]
projection = pd.DataFrame(index=years)

for scenario, params in scenarios.items():
    revenue = [base_revenue * (1 + params["growth"]) ** y for y in years]
    fcf = [rev * params["fcf_margin"] for rev in revenue]
    projection[f"Revenue ({scenario})"] = revenue
    projection[f"FCF ({scenario})"] = fcf

print(projection.round(0))
```

```python
shares_outstanding = balance_sheet.loc["Ordinary Shares Number"]["2025-12-31"]
print("Shares Outstanding:", shares_outstanding)
```

```python
ads_price_usd = 42.00
usd_to_rmb = 6.72
shares_per_ads = 10

price_per_ordinary_share_rmb = (ads_price_usd / shares_per_ads) * usd_to_rmb
market_value_equity = shares_outstanding * price_per_ordinary_share_rmb

market_value_debt = balance_sheet.loc["Total Debt"]["2025-12-31"]

total_capital = market_value_equity + market_value_debt
weight_equity = market_value_equity / total_capital
weight_debt = market_value_debt / total_capital

risk_free_rate = 0.0168
equity_risk_premium = 0.0518
beta = 0.95
cost_of_equity = risk_free_rate + beta * equity_risk_premium

interest_expense = income_stmt.loc["Interest Expense"]["2025-12-31"]
pretax_income = income_stmt.loc["Pretax Income"]["2025-12-31"]
tax_provision = income_stmt.loc["Tax Provision"]["2025-12-31"]
tax_rate = tax_provision / pretax_income
cost_of_debt_pretax = interest_expense / market_value_debt
cost_of_debt_after_tax = cost_of_debt_pretax * (1 - tax_rate)

wacc = weight_equity * cost_of_equity + weight_debt * cost_of_debt_after_tax

print(f"Market Value of Equity: {market_value_equity:,.0f} RMB")
print(f"Market Value of Debt: {market_value_debt:,.0f} RMB")
print(f"Weight of Equity: {weight_equity:.2%}")
print(f"Weight of Debt: {weight_debt:.2%}")
print(f"Cost of Equity: {cost_of_equity:.2%}")
print(f"Cost of Debt (after-tax): {cost_of_debt_after_tax:.2%}")
print(f"WACC: {wacc:.2%}")
```

```python
wacc = 0.0485
terminal_growth = 0.03

results = {}
for scenario in ["Bear", "Base", "Bull"]:
    fcf_series = projection[f"FCF ({scenario})"]

    pv_fcf = sum(fcf_series.iloc[i] / (1 + wacc) ** (i + 1) for i in range(len(fcf_series)))

    terminal_value = fcf_series.iloc[-1] * (1 + terminal_growth) / (wacc - terminal_growth)
    pv_terminal_value = terminal_value / (1 + wacc) ** len(fcf_series)

    enterprise_value = pv_fcf + pv_terminal_value

    results[scenario] = {
        "PV of FCF": pv_fcf,
        "Terminal Value": terminal_value,
        "PV of Terminal Value": pv_terminal_value,
        "Enterprise Value": enterprise_value,
        "TV % of EV": pv_terminal_value / enterprise_value * 100,
    }

results_df = pd.DataFrame(results).T
print(results_df.round(0))
```

```python
cost_of_debt_pretax_assumed = risk_free_rate + 0.023  # ~4.0% pretax — risk-free + ~230bps corporate spread
cost_of_debt_after_tax_assumed = cost_of_debt_pretax_assumed * (1 - tax_rate)

wacc_corrected = weight_equity * cost_of_equity + weight_debt * cost_of_debt_after_tax_assumed

print(f"Assumed Pretax Cost of Debt: {cost_of_debt_pretax_assumed:.2%}")
print(f"Assumed After-Tax Cost of Debt: {cost_of_debt_after_tax_assumed:.2%}")
print(f"Corrected WACC: {wacc_corrected:.2%}")
```

```python
import numpy as np

wacc_range = np.arange(0.045, 0.075, 0.005)
growth_range = np.arange(0.02, 0.04, 0.005)

base_fcf_series = projection["FCF (Base)"]

sensitivity = pd.DataFrame(index=[f"{w:.1%}" for w in wacc_range],
                            columns=[f"{g:.1%}" for g in growth_range])

for w in wacc_range:
    pv_fcf = sum(base_fcf_series.iloc[i] / (1 + w) ** (i + 1) for i in range(len(base_fcf_series)))
    for g in growth_range:
        if g >= w:
            sensitivity.loc[f"{w:.1%}", f"{g:.1%}"] = None
            continue
        tv = base_fcf_series.iloc[-1] * (1 + g) / (w - g)
        pv_tv = tv / (1 + w) ** len(base_fcf_series)
        sensitivity.loc[f"{w:.1%}", f"{g:.1%}"] = pv_fcf + pv_tv

print(sensitivity)
```

```python
adr_structural_risk_premium = 0.03  # VIE/delisting/repatriation risk not captured by industry beta

cost_of_equity_adjusted = risk_free_rate + beta * equity_risk_premium + adr_structural_risk_premium
wacc_final = weight_equity * cost_of_equity_adjusted + weight_debt * cost_of_debt_after_tax_assumed

print(f"Adjusted Cost of Equity: {cost_of_equity_adjusted:.2%}")
print(f"Final WACC: {wacc_final:.2%}")
```

```python
terminal_growth = 0.03
wacc = wacc_final

ev_results = {}
for scenario in ["Bear", "Base", "Bull"]:
    fcf_series = projection[f"FCF ({scenario})"]
    pv_fcf = sum(fcf_series.iloc[i] / (1 + wacc) ** (i + 1) for i in range(len(fcf_series)))
    terminal_value = fcf_series.iloc[-1] * (1 + terminal_growth) / (wacc - terminal_growth)
    pv_terminal_value = terminal_value / (1 + wacc) ** len(fcf_series)
    ev_results[scenario] = pv_fcf + pv_terminal_value

cash_and_equivalents = balance_sheet.loc["Cash And Cash Equivalents"]["2025-12-31"]
total_debt = balance_sheet.loc["Total Debt"]["2025-12-31"]
net_debt = total_debt - cash_and_equivalents

print(f"Cash and Equivalents: {cash_and_equivalents:,.0f} RMB")
print(f"Net Debt: {net_debt:,.0f} RMB")

final_results = {}
for scenario, ev in ev_results.items():
    equity_value = ev - net_debt
    price_per_ordinary_share_rmb = equity_value / shares_outstanding
    price_per_ads_rmb = price_per_ordinary_share_rmb * shares_per_ads
    price_per_ads_usd = price_per_ads_rmb / usd_to_rmb
    final_results[scenario] = {
        "Enterprise Value (RMB)": ev,
        "Equity Value (RMB)": equity_value,
        "Implied Price per ADS (USD)": price_per_ads_usd,
    }

final_df = pd.DataFrame(final_results).T
print(final_df.round(2))
```

```python
stress_case = {
    "growth": -0.02,
    "fcf_margin": 0.18,
    "terminal_growth": 0.015,
}

revenue_stress = [base_revenue * (1 + stress_case["growth"]) ** y for y in years]
fcf_stress = [rev * stress_case["fcf_margin"] for rev in revenue_stress]
projection["Revenue (Stress)"] = revenue_stress
projection["FCF (Stress)"] = fcf_stress

print(projection[["Revenue (Stress)", "FCF (Stress)"]].round(0))
```

```python
scenario_terminal_growth = {
    "Stress": 0.015,
    "Bear": 0.03,
    "Base": 0.03,
    "Bull": 0.03,
}

final_results = {}
for scenario in ["Stress", "Bear", "Base", "Bull"]:
    fcf_series = projection[f"FCF ({scenario})"]
    g = scenario_terminal_growth[scenario]
    pv_fcf = sum(fcf_series.iloc[i] / (1 + wacc) ** (i + 1) for i in range(len(fcf_series)))
    terminal_value = fcf_series.iloc[-1] * (1 + g) / (wacc - g)
    pv_terminal_value = terminal_value / (1 + wacc) ** len(fcf_series)
    enterprise_value = pv_fcf + pv_terminal_value
    equity_value = enterprise_value - net_debt
    price_per_ordinary_share_rmb = equity_value / shares_outstanding
    price_per_ads_rmb = price_per_ordinary_share_rmb * shares_per_ads
    price_per_ads_usd = price_per_ads_rmb / usd_to_rmb
    final_results[scenario] = {
        "Enterprise Value (RMB)": enterprise_value,
        "Equity Value (RMB)": equity_value,
        "Implied Price per ADS (USD)": price_per_ads_usd,
    }

final_df = pd.DataFrame(final_results).T
print(final_df.round(2))
```

```python
long_term_debt_payments = cash_flow.loc["Long Term Debt Payments"]["2025-12-31"]
short_term_debt_payments = cash_flow.loc["Short Term Debt Payments"]["2025-12-31"]
total_debt_repayment = abs(long_term_debt_payments) + abs(short_term_debt_payments)

lease_pct_of_debt = balance_sheet.loc["Capital Lease Obligations"]["2025-12-31"] / balance_sheet.loc["Total Debt"]["2025-12-31"]
estimated_lease_principal_repayment = total_debt_repayment * lease_pct_of_debt

raw_fcf = cash_flow.loc["Free Cash Flow"]["2025-12-31"]
lease_adjusted_fcf = raw_fcf - estimated_lease_principal_repayment

revenue_2025 = income_stmt.loc["Total Revenue"]["2025-12-31"]
raw_fcf_margin = raw_fcf / revenue_2025 * 100
lease_adjusted_fcf_margin = lease_adjusted_fcf / revenue_2025 * 100

print(f"Total Debt Repayment (CF Financing): {total_debt_repayment:,.0f} RMB")
print(f"Lease % of Total Debt: {lease_pct_of_debt:.1%}")
print(f"Estimated Lease Principal Repayment: {estimated_lease_principal_repayment:,.0f} RMB")
print(f"Raw FCF: {raw_fcf:,.0f} RMB ({raw_fcf_margin:.2f}% margin)")
print(f"Lease-Adjusted FCF: {lease_adjusted_fcf:,.0f} RMB ({lease_adjusted_fcf_margin:.2f}% margin)")
```

```python
peer_manual_v2 = {}
for ticker, name in peers.items():
    t = yf.Ticker(ticker)
    info = t.info
    fin_currency = info.get("financialCurrency")
    market_cap = info.get("marketCap")

    bs = t.balance_sheet
    total_debt = bs.loc["Total Debt"].iloc[0] if "Total Debt" in bs.index else None
    cash = bs.loc["Cash And Cash Equivalents"].iloc[0] if "Cash And Cash Equivalents" in bs.index else None
    ebitda = info.get("ebitda")

    if fin_currency == "CNY" and info.get("currency") == "USD":
        total_debt_usd = total_debt / usd_to_rmb if total_debt is not None else None
        cash_usd = cash / usd_to_rmb if cash is not None else None
        ebitda_usd = ebitda / usd_to_rmb if ebitda is not None else None
    else:
        total_debt_usd = total_debt
        cash_usd = cash
        ebitda_usd = ebitda

    ev_correct = (market_cap + total_debt_usd - cash_usd) if None not in (market_cap, total_debt_usd, cash_usd) else None
    ev_ebitda_correct = (ev_correct / ebitda_usd) if ev_correct and ebitda_usd else None

    peer_manual_v2[ticker] = {
        "Financial Currency": fin_currency,
        "Market Cap (USD)": market_cap,
        "Total Debt (USD)": total_debt_usd,
        "Cash (USD)": cash_usd,
        "EBITDA (USD)": ebitda_usd,
        "EV (recalculated)": ev_correct,
        "EV/EBITDA (recalculated)": ev_ebitda_correct,
    }

peer_manual_v2_df = pd.DataFrame(peer_manual_v2).T
print(peer_manual_v2_df)
```
